```python
#Lists can contain strings, floats, and integers

# Sample List

sample = ["The Bodyguard", 7.0, 1992, [1, 2], ("A", 1)]
print(sample)
```

```python
# List slicing
# Sample List

L = ["The Bodyguard", 7.0,1992,"BG",1]
print(L[3:5])
```

```python
# Method extend

L = [ "The Bodyguard", 7.0]
print(L)
L.extend(['pop', 10])
print(L)
```

```python
# Use append to add elements to list

L = [ "The Bodyguard", 7.0]
print(L)
L.append(['pop', 10])
print(L)
L.append(['a','b'])
print(L)
```

```python
# Change the element based on the index

A = ["disco", 10, 1.2]
print('Before change:', A)
A[0] = 'hard rock'
print('After change:', A)
```

```python
# Delete the element based on the index

print('Before change:', A)
del(A[0])
print('After change:', A)
```

```python
# Split the string, default is by space

'hard rock'.split()
```

```python
# Split the string by comma

'A,B,C,D'.split(',')
```

```python
# Copy (copy by reference) the list A

A = ["hard rock", 10, 1.2]
B = A
print('A:', A)
print('B:', B)

# Examine the copy by reference

print('B[0]:', B[0])
A[0] = "banana"
print('B[0]:', B[0])
```

```python
# Clone (clone by value) the list A

B = A[:]
B

print('B[0]:', B[0])
A[0] = "hard rock"
print('B[0]still ==>', B[0])
```
