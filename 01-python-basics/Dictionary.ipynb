```python
# Create a sample dictionary

release_year_dict = {"Thriller": "1982", "Back in Black": "1980", \
                    "The Dark Side of the Moon": "1973", "The Bodyguard": "1992", \
                    "Bat Out of Hell": "1977", "Their Greatest Hits (1971-1975)": "1976", \
                    "Saturday Night Fever": "1977", "Rumours": "1977"}
release_year_dict
```

```python
# Get value by key

release_year_dict['The Bodyguard']
```

```python
# Get all the keys in dictionary

release_year_dict.keys()
```

```python
# Get all the values in dictionary

release_year_dict.values()
```

```python
# Append value with key into dictionary

release_year_dict['Graduation'] = '2007'
release_year_dict
```

```python
"""
Quiz

a) In the dictionary soundtrack_dic what are the keys ?
b) In the dictionary soundtrack_dic what are the values ?


The Albums Back in Black, The Bodyguard and Thriller have the following music recording sales in millions 50, 50 and 65 respectively:

a) Create a dictionary album_sales_dict where the keys are the album name and the sales in millions are the values.
b) Use the dictionary to find the total sales of Thriller:
c) Find the names of the albums from the dictionary using the method keys():
d) Find the values of the recording sales from the dictionary using the method values:
"""

# Question sample dictionary

soundtrack_dic = {"The Bodyguard":"1992", "Saturday Night Fever":"1977"}

keys = soundtrack_dic.keys()
print(keys)
values = soundtrack_dic.values()
print(values)


album_sales_dict = {"Back in Black":50, "The Bodyguard":50, "Thriller":65}
print(album_sales_dict)
sale = album_sales_dict["Thriller"]
print(f"total sales of Thriller is: {sale}")
print(album_sales_dict.keys())
print(album_sales_dict.values())







"""
Scenario
The inventory store scenario project utilizes a dictionary-based approach to develop a robust system for managing and tracking inventory in a retail store.
Note:- You will be working with two product details.

Task-1 Create an empty dictionary
First you need to create an empty dictionary, where you will be storing the product details.


Task-2 Store the first product details in variable
Product Name= Mobile phone
Product Quantity= 5
Product price= 20000
Product Release Year= 2020

Task-3 Add the details in inventory
Task-4 Store the second product details in a variable.
Task-5 Add the item detail into the inventory.
Task-6 Display the Products present in the inventory
Task-7 Check if ProductNo1_releaseYear and ProductNo2_releaseYear is in the inventory
Task-8 Delete release year of both the products from the inventory
"""
print("Hello")
inventory = {}
print(inventory)
inventory["Product Name"] = "Mobile phone"
inventory["Product Quantity"] = 5
inventory["Product Price"] = 20000
inventory["Product Release Year"] = 2020
print("Products present in the inventory:")
display(inventory)

has_product1_release = 'Product Release Year' in inventory
has_product2_release = 'ProductNo2_releaseYear' in inventory

print(f"Is Product 1 Release Year in inventory? {has_product1_release}")
print(f"Is Product 2 Release Year in inventory? {has_product2_release}")

if 'Product Release Year' in inventory:
    del inventory['Product Release Year']

if 'ProductNo2_releaseYear' in inventory:
    del inventory['ProductNo2_releaseYear']

print("Inventory after deleting release years:")
display(inventory)
```
