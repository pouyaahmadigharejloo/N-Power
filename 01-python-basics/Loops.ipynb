While Loop

```python
# While Loop Example

dates = [1982, 1980, 1973, 2000]

i = 0
year = dates[0]

while(year != 1973):
    print(year)
    i = i + 1
    year = dates[i]


print("It took ", i ,"repetitions to get out of loop.")
```

Controlling Loop Execution with break and continue

The break statement

```python
for num in range(1, 10):
    if num == 5:
        print("Breaking the loop at:", num)
        break
    print(num)
```

The continue statement

```python
for num in range(1, 6):
    if num == 3:
        continue
    print(num)
```

```python
# Example 3: Using break and continue in a while loop
count = 0
while count < 10:
    count += 1
    if count == 3:
        continue  # skip printing 3
    if count == 8:
        break     # stop the loop when count is 8
    print(count)

    # Note: Here, 3 is skipped due to continue, and the loop stops when count reaches 8 due to break.
```

```python
"""
Write a Python program using a for loop that prints numbers from 1 to 15 but:

Skips multiples of 3
Stops the loop if the number is greater than 12
"""
for i in range(1, 16):
    if i % 3 == 0:
        continue  # skip multiples of 3
    if i > 12:
        break     # stop if number > 12
    print(i)
```
