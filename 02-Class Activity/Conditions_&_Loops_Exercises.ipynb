```python
#Q6 Print the follwoing pattern
#*
#**
#***
#****

i = "*"

while len(i) < 5:
    print(i)
    i = i + "*"
```

```python
#Q7 Reverse the following list using for loop

list1=[90,70,50,60]

for num in list1[::-1]:
    print(num)
```

```python
#Q8 Display numbers from -10 to -1 using for loop [hint use range method]

for num in range(-10,0):
  print(f"({num}", end=")")
```
